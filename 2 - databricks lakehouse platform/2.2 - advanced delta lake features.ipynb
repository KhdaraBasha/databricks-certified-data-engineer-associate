```python
%sql
-- Step 1: Switch to the demo catalog
USE CATALOG databricks_demo;
```

```python
%sql
-- Step 2: View history of employees Delta table
DESCRIBE HISTORY employees
```

```python
%sql
-- Step 3: Query employees table at version 4
SELECT *
FROM employees VERSION AS OF 4
```

```python
%sql
-- Step 4: Query employees table at version 6
SELECT *
FROM employees VERSION AS OF 6
```

```python
%sql
-- Step 5: Query employees table using versioned syntax
SELECT *
FROM employees@v6
```

```python
%sql
-- Step 6: Delete all records from employees table
DELETE FROM employees;
```

```python
%sql
-- Step 7: Query employees table after deletion
SELECT * FROM employees
```

```python
%sql
DESCRIBE HISTORY employees
```

```python
%sql
RESTORE TABLE employees TO VERSION AS OF 6;
    
-- Step 8: Query employees table after restore
SELECT * FROM employees;
```

```python
%sql
DESCRIBE HISTORY employees;
```
