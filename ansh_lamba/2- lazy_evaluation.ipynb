# **_Lazy Evaluation in Databricks_**

Databricks, built on Apache Spark, uses lazy evaluation for transformations. This means operations like `select`, `filter`, and `map` are not executed immediately. Instead, Spark builds a logical plan and only executes the computation when an action (e.g., `show()`, `collect()`, `count()`) is called. This approach optimizes query execution and resource usage.

**Example:**
python
df = spark.read.csv("data.csv")
filtered = df.filter(df["age"] > 30)  # Transformation (lazy)
filtered.show()  # Action (triggers execution)

```python
# =============================================================================
# LAZY EVALUATION DEMONSTRATION
# =============================================================================
# This cell demonstrates how Spark's lazy evaluation works. Transformations
# (select, filter, map, etc.) are not executed immediately - they only build
# a logical execution plan. Actual computation happens when an action is called.
# =============================================================================

# Sample employee data with 11 records
# Structure: (id, first_name, last_name, email, year, salary, country)
data = [
  (100, "james", "smith", "s.james@gmail.com" , 1997, 30000, "USA"),
  (101, "michael", "rose",  "michael@roses.com", 2000, 40000, "USA"),
  (102, "robert", "williams", "robert.williams@yahoo.com", 1999, 40000, "USA"),
  (103, "mary", "jones", "mary.jones@gmail.com", 2002, 60000, "USA"),
  (104, "susan", "martin", "susan.martin@yahoo.com", 1998, 65000, "USA" ),
  (105, "david", "brown", "david.brown@gmail.com", 2005, 70000, "USA" ),
  (106, "julie", "davis", "julie.davis@yahoo.com", 2002, 75000, "USA"),
  (107, "jennifer", "thomas", "jennifer.thomas@gmail.com", 2004, 80000, "USA"),
  (108, "charles", "lewis", "charles.lewis@yahoo.com", 2001, 85000, "USA"),
  (109, "doris", "king", "doris.king@gmail.com", 2003, 90000, "USA"),
  (110, "mark", "williams", "mark.williams@yahoo.com", 2000, 95000, "USA")
]

# Define the schema for the DataFrame
# Explicitly declaring column names and data types ensures type safety
schema = "id INTEGER, first_name STRING, last_name STRING, email STRING, year INTEGER, salary INTEGER, country STRING"

# Create a DataFrame from the data list and schema
# This is a DataFrame creation operation, not a transformation
df = spark.createDataFrame(data, schema)

# =============================================================================
# TRANSFORMATION - 1 (LAZY): Select specific columns
# =============================================================================
# The select() operation is a TRANSFORMATION - it is NOT executed immediately
# Spark simply records this operation in the execution plan (DAG)
# No data is actually processed or filtered at this point
df = df.select("id", "first_name", "last_name", "email", "salary")

# =============================================================================
# TRANSFORMATION - 2 (LAZY): Filter rows where id >= 105
# =============================================================================
# The filter() operation is also a TRANSFORMATION - it is NOT executed yet
# Spark adds this filter condition to the logical plan
# Still no actual computation has occurred - we're just building the plan
# When an action is triggered, Spark will optimize and execute both transformations together
df = df.filter(df.id >= 105)
```

```python
# =============================================================================
# ACTION: display() - TRIGGERS EXECUTION
# =============================================================================
# The display() function is an ACTION in Databricks
# This is the moment when Spark actually executes all the transformations:
#   1. Reads the data from the DataFrame
#   2. Applies the select() to choose only 5 columns
#   3. Applies the filter() to keep only rows where id >= 105
#   4. Returns and displays the result
#
# Before this action, Spark only had a logical plan of what to do.
# Now it optimizes the plan and executes it efficiently.
# Result: Only 6 rows (ids 105-110) with 5 columns will be displayed
# =============================================================================
display(df)
```

```python
# =============================================================================
# ACTION: show() - ALSO TRIGGERS EXECUTION
# =============================================================================
# The show() method is another ACTION in Spark
# Just like display(), this triggers the execution of all transformations
# 
# Key differences between show() and display():
#   - show(): Standard Spark action, prints results in table format to console
#             Default shows 20 rows, can customize with show(n)
#   - display(): Databricks-specific, provides interactive table with sorting,
#                filtering, and visualization options
#
# Since the DataFrame 'df' hasn't changed since the previous cell,
# this will show the same 6 filtered rows (ids 105-110)
# 
# Note: Each action independently triggers execution. If we had multiple actions,
# Spark would re-execute the transformations for each one (unless caching is used)
# =============================================================================
df.show()
```
