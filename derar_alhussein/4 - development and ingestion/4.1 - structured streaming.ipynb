# Structured Streaming in PySpark

Structured Streaming is a scalable and fault-tolerant stream processing engine built on the Spark SQL engine. It enables real-time processing of continuous data streams with the same semantics as batch processing.

**Key Concepts:**
* Data is processed as an unbounded table using the same API as static DataFrames.
* Supports sources like files, Kafka, socket, Delta Lake, and custom sources.

**Key Methods:**
* `.readStream` and `.writeStream` for defining streaming DataFrames and outputting results.
* Streaming queries run continuously and can be managed with methods like `start()`, `stop()`, and `awaitTermination()`.
* Common triggers include processing time, once, and available-now modes.
* Supports windowed aggregations for real-time summarization.

**Typical Usage:**
```python
stream_df = spark.readStream.format('delta').table('events_table')
query = stream_df.writeStream.format('console').start()
query.awaitTermination()
```

**Management and Monitoring:**
* Check status with `query.status` and `query.lastProgress`
* List active queries with `spark.streams.active`

See Databricks documentation for advanced features like watermarking, stateful operations, fault tolerance, and integrating with Delta Lake.

# Detailed Explanation of `readStream` and `writeStream` in PySpark

## `readStream`
Structured streaming uses `spark.readStream` to define a streaming DataFrame. This method supports sources such as files, Kafka, Delta Lake, socket, and more.

**Common usage:**
```python
stream_df = spark.readStream.format('parquet').load('/path/to/data')
stream_df = spark.readStream.format('kafka').option('kafka.bootstrap.servers', 'host:port').option('subscribe', 'topic').load()
```

**Important Parameters:**
* `format`: Source type (`parquet`, `delta`, `kafka`, `csv`, etc.)
* `option`: Specific settings like file path, Kafka topic, host, schema, etc.
* `schema`: Defines schema for sources without schema inference (e.g., JSON, CSV, socket)

**Returns:** Streaming DataFrame representing ongoing data.

## `writeStream`
This method is used to output results from streaming DataFrames. It controls output mode, destination, checkpointing, and trigger.

**Common usage:**
```python
query = stream_df.writeStream.outputMode('append').format('console').start()
query.awaitTermination()
```

**Important Parameters:**
* `outputMode`: `'append'`, `'complete'`, `'update'` (how data is written)
* `format`: Sink type (`console`, `parquet`, `delta`, `memory`, `foreach`)
* `option`: Output-specific settings (e.g., path, trigger time, checkpoint location)
* `queryName`: Assigns a name for management and monitoring
* `trigger`: Customizes execution timing (`processingTime`, `once`, `availableNow`)

**Returns:** Streaming query object for lifecycle management.

These two methods form the foundation for building robust streaming pipelines in PySpark.
