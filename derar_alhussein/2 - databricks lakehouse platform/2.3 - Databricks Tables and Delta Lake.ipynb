```python
# Create a managed table using Delta Lake in Databricks
# This cell creates a managed table backed by Delta format and a specified S3 location
spark.sql("""
CREATE TABLE IF NOT EXISTS my_managed_table (
  id INT,
  name STRING
)
USING DELTA
LOCATION 's3://databricks-demo-data-01/managed_data/my_managed_table'
""")
```

```python
# Insert initial records into the managed table
# This cell adds three sample rows to 'my_managed_table' using SQL insert
spark.sql("""
INSERT INTO my_managed_table VALUES
  (1, 'Alice'),
  (2, 'Bob'),
  (3, 'Charlie')
  """)
```

```python
%sql
-- Show all data in the managed table
-- This SQL query selects all rows from 'my_managed_table'
select * from my_managed_table;
```

```python
%sql
-- View Delta transaction history for managed table
-- This command displays the table's version control history
 describe history my_managed_table
```

```python
# Insert additional records into the managed table
# Demonstrates multiple insert statements for appending more rows
spark.sql("""
INSERT INTO my_managed_table VALUES
  (4, 'Dave'),
  (5, 'Eve')
  """);

# Second insertion batch
spark.sql("""
INSERT INTO my_managed_table VALUES
  (6, 'Frank'),
  (7, 'Grace')
  """);

# Third insertion batch
spark.sql("""
INSERT INTO my_managed_table VALUES
  (8, 'Henry'),
  (9, 'Ivy')
  """);
```

```python
%sql
-- View the transaction history for managed table (again)
-- Shows all changes up to the current state
describe history my_managed_table;
```

```python
%undefined
-- List files in the managed table S3 directory
-- %fs magic command to show files stored for the table
%fs ls 's3://databricks-demo-data-01/managed_data/my_managed_table'
```

```python
%undefined
-- List Delta log files for the managed table
-- Uses %fs magic to show _delta_log contents
%fs ls 's3://databricks-demo-data-01/managed_data/my_managed_table/_delta_log/'
```

```python
# Read and display a Delta transaction log file
# Demonstrates reading a Delta Lake JSON log directly using PySpark
# This cell displays a specific commit log from the Delta Lake table directory
display(spark.read.json('s3://databricks-demo-data-01/managed_data/my_managed_table/_delta_log/00000000000000000004.json'))
```

```python
%sql
-- Show extended table metadata for managed table
-- Reveals schema, file and storage details
DESCRIBE EXTENDED my_managed_table
```

```python
%sql
-- Show detailed properties for the managed table
-- Includes table format, provider, physical location, etc.
DESCRIBE DETAIL my_managed_table
```

```python
%sql
-- View Delta transaction history for managed table (redundant)
-- Repeats the describe history command
 describe history my_managed_table
```

```python
# Show all records in the managed table using Spark
# Executes a PySpark SQL query and prints results
spark.sql("select * from my_managed_table").show()
```

```python
# Update a record in managed table
# Changes the 'name' for id = 2 to 'Basha'
spark.sql(
    """
    update my_managed_table
    set name = 'Basha'
    where id = 2
    """
)
```

```python
%sql
-- Show full Delta table operation history
-- Verifies all commits and DML/DDL actions
DESCRIBE HISTORY my_managed_table
```

```python
%sql
-- Show full managed table properties and storage location
-- Displays information such as table schema, format, and storage path
DESCRIBE DETAIL my_managed_table
```

```python
# Drop managed table from metastore
# This removes the table and metadata, but does not delete underlying S3 data
spark.sql("DROP TABLE my_managed_table")
```

```python
# Create an external table referencing S3 location
# Explicit schema for Parquet external table; path must exist but can be empty initially
spark.sql("""
CREATE TABLE IF NOT EXISTS my_external_table (
  id INT,
  name STRING
)
USING PARQUET
LOCATION 's3://databricks-demo-data-01/external_data/my_external_table'
""")
```

```python
# Insert sample records into the external table
# This cell adds three rows to the table on S3 using SQL insert
spark.sql("""
INSERT INTO my_external_table VALUES
  (1, 'Alice'),
  (2, 'Bob'),
  (3, 'Charlie')
""")
```
