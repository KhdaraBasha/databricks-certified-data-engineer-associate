```python
%sql
CREATE TABLE managed_table
(width INT, length INT, height INT)
;

INSERT INTO managed_table VALUES
(10, 10, 10),
(20, 20, 20),
(30, 30, 30)
;
```

```python
%sql
CREATE TABLE external_table
(width INT, length INT, height INT)
USING DELTA
LOCATION 's3://s3-databricks-ext-tables/'
;
```

```python
%sql
INSERT INTO external_table
VALUES
(10, 10, 10),
(20, 20, 20),
(30, 30, 30)
;
    
SELECT * FROM external_table
;
```

```python
%sql
DESCRIBE EXTENDED external_table;
```
