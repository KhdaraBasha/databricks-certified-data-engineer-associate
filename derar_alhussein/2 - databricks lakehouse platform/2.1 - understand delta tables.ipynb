```python
%sql
USE CATALOG databricks_demo;
```

```python
%sql
DROP TABLE IF EXISTS employees;
CREATE TABLE employees 
(id INT, name STRING, salary INT, department STRING)
USING DELTA
;
```

```python
%sql
DESCRIBE TABLE employees;
```
