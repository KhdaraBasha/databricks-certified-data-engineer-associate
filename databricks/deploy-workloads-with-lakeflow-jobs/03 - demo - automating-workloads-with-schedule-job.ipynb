```python
# Creates a multi-task job to schedule notebooks
from databricks.sdk import WorkspaceClient
from databricks.sdk.service.jobs import Task, NotebookTask, Source

w = WorkspaceClient()

# Define the tasks
job_tasks = [
  {
    'task_name': 'ingesting_orders',
    'file_path': '3.3 - creating-order-tables',
    'depends_on': None
  },
  {
    'task_name': 'ingesting_sales',
    'file_path': '3.2 - creating-sales-table',
    'depends_on': None
  }
]

# Get the current user's directory for notebook paths
user_dir = '/Users/azurelearn2018@gmail.com/databricks-certified-data-engineer-associate/databricks/deploy-workloads-with-lakeflow-jobs'

# Create task objects
tasks = []
for task_config in job_tasks:
    notebook_path = f"{user_dir}/{task_config['file_path']}"
    
    task = Task(
        task_key=task_config['task_name'],
        notebook_task=NotebookTask(
            notebook_path=notebook_path,
            source=Source.WORKSPACE
        ),
        depends_on=[{'task_key': task_config['depends_on']}] if task_config['depends_on'] else []
    )
    tasks.append(task)

# Create the job with serverless compute
myjob = w.jobs.create(
    name='Data Ingestion Pipeline',
    tasks=tasks
)

print(f"Job created successfully with ID: {myjob.job_id}")
print(f"Job URL: {w.config.host}/#job/{myjob.job_id}")
```
